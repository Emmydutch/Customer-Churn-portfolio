## Stage 1 — Project Definition

This stage establishes the business problem, project objectives, scope, stakeholders, and success criteria. It ensures that every later analysis, visualization, model, and recommendation supports a clear business decision.

## 1.1 Executive Summary

A telecommunications company serves customers with different demographic profiles, contracts, services, payment arrangements, and levels of engagement. Some customers remain with the company, some have recently joined, and others discontinue their service. Customer losses reduce recurring revenue, weaken customer lifetime value, and may increase the cost of replacing departing customers.

This project will use customer-level data to measure churn, identify the customer groups most exposed to it, investigate the reasons customers leave, and translate the findings into practical retention actions. The final output will be an interactive decision-support dashboard supported by a transparent and reproducible analytical workflow.

## 1.2 What Customer Churn Means

In this project, **customer churn** means that a customer has ended or discontinued their relationship with the telecom company and is recorded with a churn label of `Yes` and a customer status of `Churned`. Customers labelled `No` have not churned; this group contains customers who stayed as well as customers who recently joined.

The main descriptive churn rate will be calculated as:

**Churn Rate = Number of Churned Customers ÷ Total Number of Customers × 100**

Where a metric requires a different eligible population—for example, separating newly joined customers from established customers—the denominator will be stated clearly.

## 1.3 Business Problem

The company has experienced customer attrition, but an overall churn figure alone does not reveal where the greatest retention risks exist or which interventions should be prioritized. Management needs a reliable way to distinguish high-risk customer segments, understand the commercial and service-related reasons behind churn, and direct retention resources toward actions with the greatest potential value.

Without this insight, retention campaigns may be broad, reactive, expensive, and poorly targeted. The business therefore requires an evidence-based churn intelligence solution that connects customer characteristics and service experiences with churn outcomes, customer value, and appropriate retention responses.

## 1.4 Central Business Question

> **Which customers are most likely to churn, why are they leaving, and what actions could improve retention?**

```python
yes_no_columns = [
    "Under30", "SeniorCitizen", "Married", "Dependents",
    "ReferredaFriend", "PhoneService", "MultipleLines",
    "InternetService", "OnlineSecurity", "OnlineBackup",
    "DeviceProtectionPlan", "PremiumTechSupport", "StreamingTV",
    "StreamingMovies", "StreamingMusic", "UnlimitedData",
    "PaperlessBilling", "ChurnLabel"
]

invalid_yes_no = {
    column: sorted(set(df[column].dropna()) - {"Yes", "No"})
    for column in yes_no_columns
}
invalid_yes_no = {column: values for column, values in invalid_yes_no.items() if values}

consistency_checks = pd.Series({
    "Under30 disagrees with Age": int(((df["Age"] < 30) != df["Under30"].eq("Yes")).sum()),
    "SeniorCitizen disagrees with Age >= 65": int(((df["Age"] >= 65) != df["SeniorCitizen"].eq("Yes")).sum()),
    "Dependents disagrees with dependent count": int(((df["NumberofDependents"] > 0) != df["Dependents"].eq("Yes")).sum()),
    "Referral flag disagrees with referral count": int(((df["Number_of_Referrals"] > 0) != df["ReferredaFriend"].eq("Yes")).sum()),
    "No phone service but non-zero long-distance average": int(((df["PhoneService"] == "No") & df["AvgMonthlyLongDistanceCharges"].ne(0)).sum()),
    "No internet but non-zero download usage": int(((df["InternetService"] == "No") & df["AvgMonthlyGBDownload"].ne(0)).sum()),
    "No internet but internet type present": int(((df["InternetService"] == "No") & df["InternetType"].notna()).sum()),
    "Internet customer with missing internet type": int(((df["InternetService"] == "Yes") & df["InternetType"].isna()).sum()),
    "CustomerStatus disagrees with ChurnLabel": int(((df["CustomerStatus"] == "Churned") != df["ChurnLabel"].eq("Yes")).sum()),
    "Non-churner has churn details": int(((df["ChurnLabel"] == "No") & (df["ChurnCategory"].notna() | df["ChurnReason"].notna())).sum()),
    "Churner is missing churn details": int(((df["ChurnLabel"] == "Yes") & (df["ChurnCategory"].isna() | df["ChurnReason"].isna())).sum()),
})

revenue_recalculated = (
    df["TotalCharges"]
    + df["TotalLongDistanceCharges"]
    + df["TotalExtraDataCharges"]
    - df["TotalRefunds"]
)

print("Invalid Yes/No values:", invalid_yes_no or "None")
display(consistency_checks.to_frame("FailedRecords"))
print("Maximum TotalRevenue reconciliation difference:",
      round((df["TotalRevenue"] - revenue_recalculated).abs().max(), 10))
```

## 2.8 Constant, Redundant, and High-Cardinality Fields

- `Country`, `State`, and `Quarter` each contain one value only. They provide context but no within-dataset explanatory or predictive variation.
- `Under30` and `SeniorCitizen` are exact age-derived indicators. `Dependents` and `ReferredaFriend` are likewise derivable from their count variables. Retaining both versions is helpful for reporting, but using all of them together may add redundancy to a model.
- `CustomerID` is unique and carries no intended behavioral meaning. It will be retained for lookup but excluded from predictive features.
- `City` and `ZipCode` have high cardinality relative to the dataset size. They are useful for maps and aggregated hotspot analysis, but naive one-hot encoding could create sparse, unstable model features.
- Latitude, longitude, and ZIP-code population describe service-area context rather than an individual customer's behavior and must not be interpreted as personal attributes.

## 2.9 Outcome Timing and Leakage Classification

| Classification | Variables | Use |
|---|---|---|
| Primary target | ChurnLabel | Defines the binary modeling outcome |
| Direct outcome leakage | CustomerStatus | Exclude from predictive inputs because it directly identifies churn |
| Post-churn information | ChurnCategory, ChurnReason | Use only to explain completed churn events |
| Existing model output | ChurnScore | Exclude from the new model; otherwise the project would model another model's score |
| Timing/provenance uncertain | SatisfactionScore | Use descriptively; exclude from the baseline model until it is confirmed that the score was available before churn |
| Derived commercial estimate | CLTV | Use to prioritize value; test separately in modeling only after its calculation timing is understood |
| Candidate pre-churn information | Demographics, tenure, services, contract, billing, charges, referrals, and geography | Eligible subject to preprocessing, fairness, redundancy, and availability checks |

This classification prevents target leakage and keeps the eventual model aligned with the real decision point: identifying risk before a customer has already left.

## 2.10 Dataset Limitations

2. **Single quarter:** all records are marked Q3, so quarterly trends, seasonality, and changes in churn over time cannot be measured.

4. **No churn date:** the exact timing of churn and time-to-event behavior cannot be studied.
5. **No interaction history:** calls, complaints, outages, support tickets, campaign contacts, and resolution times are unavailable.
6. **No acquisition or campaign cost:** retention profitability can only be illustrated through transparent scenarios, not measured as realized return on investment.

8. **Unverified provenance for derived scores:** ChurnScore, SatisfactionScore, and CLTV require documentation of their calculation and timing before predictive use.
9. **Observational data:** relationships may reflect confounding factors and should not be described as causal effects.
10. **Source attribution pending verification:** the immediate source is the supplied CSV; the original publisher URL and licence must be confirmed before public distribution.

# Stage 3 — Business KPI Framework

This stage defines the project's decision metrics before visualization and modeling. Each KPI has a fixed business meaning, population, formula, display convention, and interpretation so that the notebook, dashboard, and README report consistent values.

## 3.1 Measurement Principles

1. `CustomerID` is the counting unit; each customer is counted once.
2. `ChurnLabel` defines the binary churn outcome.
3. `CustomerStatus` separates non-churned customers into `Stayed` and `Joined`.
4. Rates always state their eligible denominator.
5. Stock measures, recurring monthly amounts, accumulated revenue, and estimated lifetime value are reported separately.
6. Historical revenue from churned customers is not described as revenue lost.
7. High-risk counts will be based on the project's validated model and operating threshold—not the supplied `ChurnScore`.
8. Dashboard filters must recompute numerators and denominators from the same filtered customer population.

```python
# Boolean masks create one consistent definition for every downstream KPI.
is_churned = df["ChurnLabel"].eq("Yes")
is_stayed = df["CustomerStatus"].eq("Stayed")
is_joined = df["CustomerStatus"].eq("Joined")
is_active = ~is_churned

total_customers = df["CustomerID"].nunique()
churned_customers = int(is_churned.sum())
stayed_customers = int(is_stayed.sum())
new_customers = int(is_joined.sum())
active_customers = int(is_active.sum())

kpi_values = {
    "Total customers": total_customers,
    "Churned customers": churned_customers,
    "Retained / stayed customers": stayed_customers,
    "New / joined customers": new_customers,
    "Active customers": active_customers,
    "Overall churn rate": churned_customers / total_customers,
    "Portfolio non-churn rate": active_customers / total_customers,
    "Established-customer retention rate": stayed_customers / (stayed_customers + churned_customers),
    "Average monthly charge": df["MonthlyCharge"].mean(),
    "Average tenure (months)": df["TenureinMonths"].mean(),
    "Churned-customer accumulated revenue": df.loc[is_churned, "TotalRevenue"].sum(),
    "Churned-customer monthly revenue exposure": df.loc[is_churned, "MonthlyCharge"].sum(),
    "Average customer lifetime value": df["CLTV"].mean(),
    "Churned-customer CLTV": df.loc[is_churned, "CLTV"].sum(),
}

pd.Series(kpi_values, name="Value").to_frame()
```

## 7.2 Prioritization Method

Each actionable segment is assessed using:

1. **Observed churn rate:** churned customers divided by all customers in the segment.
2. **Active-customer reach:** currently non-churned customers matching the segment.
3. **Active monthly-charge exposure:** sum of current monthly charges among those active customers.
4. **Active CLTV:** sum of supplied CLTV among active customers in the segment.
5. **Opportunity index:** observed segment churn rate multiplied by active monthly-charge exposure.

The opportunity index is a transparent ranking heuristic balancing historical churn intensity and reachable monthly commercial exposure. It is not forecast loss, a calibrated probability, or expected campaign return. Because segments overlap, opportunity values are not additive.

# Stage 8 — Churn-Driver Assessment and Statistical Validation

This stage tests whether observed churn relationships are statistically distinguishable, measures their practical strength, and determines whether each field would genuinely be available before a customer leaves. Statistical association, predictive usefulness, operational availability, and causality are treated as separate questions.

## 8.1 Analytical Framework

A credible churn driver must pass more than a significance test:

1. **Temporal availability:** could the business know the value before churn?
2. **Outcome independence:** is the field distinct from the churn label or a model already trained on it?
3. **Association strength:** is the effect large enough to matter, not merely statistically significant in a large sample?
4. **Operational meaning:** can the business interpret and potentially act on the information?
5. **Redundancy control:** does the variable duplicate another feature or simply accumulate with tenure?
6. **Fairness:** could use of demographic information create inequitable treatment?
7. **Causal restraint:** does the evidence show association only, or support an intervention claim?

No variable in this observational snapshot is declared a proven causal driver. The term *potential driver* refers to a plausible pre-churn factor supported by descriptive and statistical evidence.

## 8.2 Statistical Methods

- **Categorical variables:** Pearson chi-square tests assess dependence with churn; bias-corrected Cramér's V measures effect strength.
- **Numerical variables:** two-sided Mann–Whitney U tests compare churned and non-churned distributions without assuming normality; rank-biserial correlation measures direction and effect size.
- **Multiple testing:** Benjamini–Hochberg false-discovery-rate correction is applied within each test family at 5%.
- **Effect interpretation:** Cramér's V below 0.10 is labelled negligible, 0.10–0.29 small, 0.30–0.49 moderate, and 0.50+ strong. Absolute rank-biserial correlation uses the same practical bands.
- **Assumption review:** categorical tests record expected cell counts; sparse results require caution.
- **Redundancy:** Spearman correlations of at least 0.70 are flagged for feature-selection review.

Threshold labels are practical conventions rather than universal laws. Sample size, business impact, and field availability remain part of interpretation.

```python
from src.statistical_validation import (
    build_feature_governance,
    categorical_churn_associations,
    export_validation_outputs,
    numerical_churn_associations,
    numerical_redundancy_report,
)

categorical_associations = categorical_churn_associations(eda_df)
numerical_associations = numerical_churn_associations(eda_df)
feature_governance = build_feature_governance(eda_df)
numerical_redundancy = numerical_redundancy_report(eda_df)
validation_files = export_validation_outputs(FEATURE_DATA_PATH)

for output_name, output_path in validation_files.items():
    print(f"{output_name}: {output_path}")
```

# Stage 10 — Predictive Modeling and Model Evaluation

This stage develops a reproducible churn-probability model using only plausibly available pre-churn operational information. A dummy baseline, logistic regression, decision tree, random forest, and gradient-boosting model are evaluated on identical stratified training folds. Model selection balances ranking performance with interpretability rather than selecting complexity automatically.

## 10.1 Modeling Objective

The modeling target is `churn_flag`, where 1 represents churn and 0 represents no churn. The model estimates the probability that a customer belongs to the churn class based on information assumed to be available before departure.

The objective is not to reproduce the supplied `ChurnScore`. It is to build an independent, auditable pipeline that can rank customers for retention review. Model probability represents statistical risk under the available data—not certainty that a customer will leave or evidence that a proposed intervention will work.

## 10.2 Leakage-Safe Feature Set

The baseline uses 24 operational predictors: eight numerical and sixteen categorical fields.

### Numerical

- Number of dependents and referrals
- Tenure
- Average monthly long-distance charge and download usage
- Current monthly charge
- Total refunds and extra-data charges

### Categorical

- Marital status and offer
- Phone, multiple lines, and internet type
- Security, backup, device protection, premium support, streaming, and unlimited-data services
- Contract, paperless billing, and payment method

Excluded fields include the target and status, churn reasons/categories, supplied ChurnScore, satisfaction, CLTV, identifiers, constant fields, geography, sensitive age/gender indicators, descriptive risk rules, and redundant accumulated revenue measures. Demographic fields remain available for later subgroup auditing but do not drive baseline predictions.

## 10.3 Reproducible Experimental Design

- **Holdout:** 20% of customers are reserved as an untouched test set.
- **Training:** 80% is used for model development.
- **Stratification:** both sets preserve the 26.54% churn prevalence.
- **Cross-validation:** five shuffled stratified folds with random seed 42.
- **Preprocessing:** median imputation and standardization for numerical fields; most-frequent imputation and one-hot encoding for categorical fields.
- **Leakage prevention:** all preprocessing is inside the scikit-learn pipeline and is refitted within every training fold.
- **Selection metric:** average precision, also called PR-AUC, because it focuses on identifying the minority churn class.
- **Selection rule:** choose the simplest model within 0.02 mean cross-validated PR-AUC of the best candidate.
- **Final test:** the selected pipeline is evaluated on the holdout set only after the selection rule is applied.

The split contains 5,634 training customers and 1,409 test customers. The test set contains 374 churners and 1,035 non-churners.

```python
from sklearn.metrics import precision_recall_curve, roc_curve

from src.modeling import (
    CATEGORICAL_FEATURES,
    MODEL_FEATURES,
    NUMERIC_FEATURES,
    train_model_suite,
)

model_result = train_model_suite(FEATURE_DATA_PATH)
selected_model_name = model_result["selected_model_name"]
cv_model_results = model_result["cv_results"]
holdout_results = model_result["test_results"]
selected_importance = model_result["feature_importance"]

print(model_result["selection_rationale"])
print(f"Training customers: {model_result['train_size']:,}")
print(f"Test customers: {model_result['test_size']:,}")
```

# Stage 12 — Model Explanation

This stage explains what the selected logistic model has learned globally and why it assigns a particular probability to an individual customer. Model behavior is explained separately from business causality: a feature can influence a prediction without proving that changing the feature will change churn.

## 12.1 Explanation Strategy

The explanation framework uses four complementary views:

1. **Holdout permutation importance:** measures the average loss in PR-AUC when one original input is shuffled. This shows which inputs the fitted model relies on for ranking unseen customers.
2. **Logistic coefficients:** show conditional direction and odds multipliers after preprocessing, relative to reference categories and other included fields.
3. **Marginal prediction profiles:** vary one field across the holdout background while keeping all others unchanged, then average the resulting probabilities.
4. **Exact local decomposition:** separates each customer's log-odds from a mean-transformed portfolio reference into additive feature contributions.

The SHAP package is not installed in the project environment. For logistic regression, exact coefficient-based additive decomposition provides a transparent local explanation without an approximation or additional dependency.

## 12.3 Logistic Effects and Conditional Direction

A positive coefficient increases modeled churn log-odds relative to the field's reference or standardized baseline; a negative coefficient reduces them. Exponentiating a coefficient gives a conditional odds multiplier.

For numerical variables, the multiplier represents one training-standard-deviation increase. For categorical terms, it represents the listed category relative to the encoder's omitted reference. The model contains 36 transformed terms from 24 original inputs.

Coefficients must be interpreted conditionally. For example, bivariate EDA shows fiber optic has high churn, while the model's fiber coefficient can turn negative after holding monthly charge, contract, support, streaming, and other correlated service fields constant. This is statistical suppression, not evidence that fiber service prevents churn. Similar reversals can occur for marital status or phone configuration. Business recommendations should therefore combine EDA, validated associations, and model explanations rather than reading one coefficient causally.

## 12.5 How an Individual Prediction Is Explained

The production model's transformed feature means define a reference log-odds value. Its sigmoid is **10.02%**; this is the probability at the mean transformed feature vector, not the arithmetic average customer probability.

For each customer:

1. Every transformed value is centered against the portfolio background mean.
2. The centered value is multiplied by its logistic coefficient.
3. One-hot terms are grouped back to their original business feature.
4. Positive contributions raise risk relative to the reference; negative contributions lower it.
5. The reference log-odds plus all grouped contributions exactly reconstruct the customer's probability.

This explains why the model produced a score. It does not say that changing the largest contribution will necessarily change the customer's behavior.

## 13.4 Simulation Formulas

For each scenario:

- **Expected reached customers** = eligible customers × reach rate
- **Expected offer acceptors** = eligible × reach rate × acceptance rate
- **Expected customers saved** = sum of churn probabilities × reach × acceptance × incremental save rate
- **Contact cost** = eligible customers × contact cost per target
- **Offer cost** = sum of intervention costs × reach × acceptance × cost multiplier
- **Expected retained gross margin** = sum(probability × monthly charge) × reach × acceptance × save rate × retained months × gross-margin rate
- **Estimated net benefit** = retained gross margin − contact cost − offer cost
- **Estimated ROI** = net benefit ÷ total campaign cost

The incremental save rate must represent additional retention caused by treatment, not the raw percentage of contacted customers who remain. It can only be measured credibly through a treatment/control design.

## 13.8 Sensitivity and Break-Even Insight

With other base assumptions fixed, a 15% incremental save rate produces negative net benefit across acceptance rates from 30% to 70%. At a 25% save rate, the campaign becomes positive across the tested acceptance range.

Higher acceptance can reduce net benefit when the save rate is too low because more customers receive an offer without enough incremental retention value. This demonstrates why offer acceptance alone is not a sufficient campaign success metric. Incremental saves and retained margin must be measured.

## 13.9 Recommended Pilot Design

1. Select one or two operationally feasible segments, such as early fiber assurance and month-to-month support bundles.
2. Randomly assign eligible customers within each segment to treatment and business-as-usual control groups.
3. Predefine reach, acceptance, incremental retention, cost, gross-margin, and customer-experience metrics.
4. Measure retention at 30, 90, 180, and 365 days where possible.
5. Estimate incremental lift as treatment retention minus control retention, not raw treatment retention.
6. Monitor complaint rates, plan downgrades, fairness, and unintended customer harm.
7. Update scenario assumptions with observed results before scaling.
8. Preserve a control population so model targeting and treatment impact can be evaluated separately.

The model determines whom to consider contacting; the experiment determines whether the intervention works.

## 13.10 Decision Guidance

- Do not launch the optimistic scenario as a forecast; use it as an upper sensitivity case.
- Treat the base scenario as a planning hypothesis requiring pilot validation.
- Use the conservative scenario to understand downside exposure and stop-loss conditions.
- Prioritize actions that solve service or support problems before relying on discounts.
- Protect high-value customers without denying benefits to lower-value customers unfairly.
- Do not contact customers solely because of age or another protected characteristic.
- Report campaign results net of contact and offer costs.
- Maintain intervention-level treatment/control results for continuous learning.

# Stage 14 — Prescriptive Retention Decision Centre

Prediction identifies customers with elevated pre-churn risk. The prescriptive layer asks a different question:

> Given limited budget and contact capacity, which economically positive interventions should be prioritized?

For each eligible active customer, the policy estimates probability-weighted incremental saves, retained gross margin, expected contact and offer costs, net benefit, and ROI. It excludes non-positive expected-value cases, ranks the remainder by the selected objective, and admits customers while budget and capacity remain.

The policy is deterministic and auditable, but it is **not a causal uplift model**. Its incremental save rate is an explicit assumption that must be replaced with treatment-versus-control evidence from a campaign pilot.


```python
from src.decision_engine import (
    score_retention_decisions,
    select_decision_portfolio,
    summarize_decision_portfolio,
)

decision_assumptions = {
    "reach_rate": 0.75,
    "acceptance_rate": 0.50,
    "incremental_save_rate": 0.35,
    "retention_horizon_months": 12,
    "gross_margin_rate": 0.70,
    "contact_cost_per_target": 4.0,
    "offer_cost_multiplier": 1.0,
}
decision_scores = score_retention_decisions(campaign_population, **decision_assumptions)
selected_decisions, ranked_decisions = select_decision_portfolio(
    decision_scores,
    budget=15_000,
    capacity=500,
    objective="Maximize expected net benefit",
)
decision_summary = summarize_decision_portfolio(selected_decisions, ranked_decisions)
display(pd.Series(decision_summary, name="Default decision policy"))

decision_actions = (
    selected_decisions.groupby("primary_intervention", as_index=False)
    .agg(
        customers=("customer_id", "size"),
        expected_cost=("expected_campaign_cost", "sum"),
        expected_saves=("expected_incremental_saves", "sum"),
        expected_net_benefit=("expected_net_benefit", "sum"),
    )
    .sort_values("expected_net_benefit", ascending=False)
)
display(decision_actions.round(2))

fig = px.bar(
    decision_actions.sort_values("expected_net_benefit"),
    x="expected_net_benefit",
    y="primary_intervention",
    orientation="h",
    color="expected_saves",
    color_continuous_scale="Blues",
    title="Budget-Constrained Expected Net Benefit by Intervention",
)
fig.update_xaxes(title="Expected net benefit ($)", tickprefix="$")
fig.update_yaxes(title=None)
fig.update_layout(template="plotly_white", height=520)
fig.show()
```

Under the documented default assumptions, the policy selects up to 500 positive expected-value customers within a $15,000 expected budget. It connects churn probability to next-best actions without exposing customer identifiers in the public decision summary.

### Responsible decision boundary

- Do not treat predicted risk as proof that a customer will churn.
- Do not claim expected saves or net benefit as realized results.
- Do not use age, gender, or another protected characteristic as a targeting objective.
- Validate incremental lift through randomized treatment and control groups.
- Monitor selection, acceptance, retention, complaints, and outcomes across customer groups.
- Keep human review and customer welfare ahead of automated ranking.
